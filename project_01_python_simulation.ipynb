```
import numpy as np
```

```
class qubit:
  #a and b are alpha and beta representation of quantum superposition probabilities
  #superposition values are generated in this case using equiprobable random values
  def __init__(self):
    self.a = np.random.uniform(0,1)
    self.b = 1 - self.a
    
  def state(self):
    if(self.a>=self.b):
      print('0')
    else:
      print('1')

     
```

```
#number of cities n. Created matrix nxn
n_cities = 12
```

```
#n x n qubit array which is a Quantum representation of a solution
d = [ [ qubit() for y in range( n_cities ) ] for x in range( n_cities ) ]
```
